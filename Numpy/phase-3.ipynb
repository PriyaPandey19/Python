```python
import numpy as np
import matplotlib.pyplot as plt
```

advance operation with business example

```python
sales_data = np.array([
    [1,150000, 180000, 220000, 250000],
    [2,120000, 140000, 160000, 190000],
    [3,200000, 230000, 260000, 300000],
    [4,180000, 210000, 240000, 270000],
    [5,160000, 1805000, 205000, 230000],
])

print("=== Zomato sales analysis ===")
print("\n Sales data shape", sales_data.shape)
print("\n Sample data for 1st 3 restaurant",sales_data[:3])
```

```python
print(np.sum(sales_data, axis=0))
yearly_total = np.sum(sales_data[:, 1:], axis=0)
print(yearly_total)

#Min sales per restaurant
min_sales = np.min(sales_data[:, 1:], axis=1)
print(min_sales)

#max sales per year
max_sales = np.max(sales_data[:, 1:], axis=0)
print(max_sales)

#average sales per restaurant
avg_sales = np.mean(sales_data[:, 1:], axis=1)
print(avg_sales)
```

```python
cumsum = np.cumsum(sales_data[:, 1:], axis=1)
print(cumsum)

plt.figure(figsize=(8,6))
plt.plot(np.mean(cumsum, axis=0))
plt.title("Average cumulative sales across all restaurants")
plt.xlabel("Years")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```

```python
vector1 = np.array([1,2,3,4,5])
vector2 = np.array([6,7,8,8,9])

print("Vector addition: ", vector1 +vector2)

print("\n Multiplication vector", vector1 * vector2)

print("\nDot Product", np.dot(vector1, vector2))

angle = np.arccos(np.dot(vector1, vector2) / (np.linalg.norm(vector1) * np.linalg.norm(vector2)))
print(angle)
```
