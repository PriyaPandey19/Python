DataRFrames A DataFrame is a table-like structure consisting of rows and columns

```python
import numpy as np
import pandas as pd
```

#DATA FRAME USING DICTANIORY

```python
data = {
    'Name': ['John', 'Anna','Peter','Linda'],
    'Age':[28, 34, 29, 42],
    'City':['New York','Paris','Berlin','London'],
    'Salary': [65000, 70000, 62000, 85000]
}
pd.DataFrame(data)
```

#DATA FRAME USING DATA LIST

```python
data_list = [
    ['John', 28,'New York',65000],
    ['Anna', 34, 'Paris' , 42000],
    ['peter',29,'Berlin',72000],
    ['Linda', 42,'London', 85000]
]
pd.DataFrame(data_list)
columns = ["Name","Age", "City", "Price"]
df2 = pd.DataFrame(data_list, columns=columns)
df2

```

#NAME AND SELECTION OF COLUMN AND INSERTION OF DATA

```python
df2[['City','Price']]
```
