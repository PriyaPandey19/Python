```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = sns.load_dataset("tips")
df
```

1.CountPlot

```python
sns.countplot(x = df['sex'], hue = df['smoker'])
```

```python
sns.barplot(x = df['sex'],y = df['total_bill'])
```

```python
sns.barplot(x = df['sex'],y = df['tip'], estimator=np.sum)
```

2.BoxPlot

```python
sns.boxplot(x ='tip', y ='day', data= df, palette='rainbow')
```

boxplot adn violnplot are used to show the distribution of categorical data. A box plot shows the distribution of quantitive data in a way facilitates comparisons between of across levels of a categorical variables. The box shows the quartiles of teh datasets while the whiskers extend to show the rest of the distribution, except teh 'outlier'.

3.ViolinePlot

```python
sns.violinplot(x= 'day', y='total_bill',data = df)
```

```python
sns.violinplot(x= 'tip', y='day',data = df)
```

4.StripPlot

The stripplot will draw a scatterplot where one variable is categorical.A strip plot can be drawn on its own, but it is good compliment to a box or violin plot in cases where you want to show all observation along with some representation of the underlying distribution.

```python
sns.stripplot(x = 'day', y='total_bill',data = df)
```

5.SwarmPlot

```python
sns.swarmplot(x='day', y='total_bill',data = df)
```

```python
sns.violinplot(x= 'tip', y='day',data = df)
sns.swarmplot(x= 'tip', y='day',data = df)
```
