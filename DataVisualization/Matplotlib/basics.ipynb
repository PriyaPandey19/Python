```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

matplotlib basics

```python
x = np.linspace(0,5,11)
x
y = x ** 3
y
```

```python
plt.plot(x,y)
plt.title("Hello")
plt.xlabel("x axis")
plt.ylabel(" Y axis")
```

subplot in matplotlib

```python
plt.subplot(2,2,1)
plt.plot(x,y)
plt.subplot(2,2,3)
plt.plot(y,x)
plt.subplot(2,2,2)
plt.plot(x**2, x**2)
plt.subplot(2,2,4)
plt.plot(x,y)


```

object oriented way

matplotlib allows the aspect ratio, DPI and figure size to be specified when the Figure object is created. You can use the figsize and dpi keywork argument.

figsize is  a tuple of the width and height of teh figure in inches dpi is the dots-per-inch

```python
fig = plt.figure(figsize=(10, 8),dpi = 100)
axis1 = fig.add_axes([0.1,0.1,.4,.8])
axis1.plot(x,y)
axis2 = fig.add_axes([0.3,0.4,.3,.3])
axis2.plot(x,y)
```

saving plots

```python
fig = plt.figure()
ax = fig.add_axes([0.1,0.1,0.8,0.8])
ax.plot(x,y)
plt.savefig("basicplot.png")
```
